## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Make a copy of original dataset
bike_df = df.copy()

# Check missing values
bike_df.isnull().sum()
````

````python
bike_df.duplicated().sum()
````

````python
bike_df.drop_duplicates(inplace=True)
````

````python
bike_df.duplicated().sum()
````

````python
bike_df['start_time'] = pd.to_datetime(bike_df['start_time'])
bike_df['end_time'] = pd.to_datetime(bike_df['end_time'])
````

````python
bike_df['year'] = bike_df['start_time'].dt.year
bike_df['month'] = bike_df['start_time'].dt.month_name()
bike_df['day'] = bike_df['start_time'].dt.day_name()
bike_df['hour'] = bike_df['start_time'].dt.hour
````

### What all manipulations have you done and insights you found?

Data Cleaning & Manipulation Performed:

• Checked the dataset structure and data types.
• Identified missing values.
• Checked and removed duplicate records.
• Converted start_time and end_time into datetime format.
• Created new features such as Year, Month, Day, and Hour from the start_time column.
• Prepared the dataset for exploratory data analysis.

Insights:

• The dataset was successfully cleaned and transformed.
• Newly created time-based variables will help analyze monthly, daily, and hourly bike usage patterns.
